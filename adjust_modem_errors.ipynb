```python
from mtpy import MTData
import numpy as np
from pathlib import Path
```

```python
md = MTData()
md.from_modem(
    r"c:\Users\jpeacock\OneDrive - DOI\Geothermal\GreatBasin\modem_inv\mina_inv_01\mina_modem_data_z03_t02_tec_04.dat"
)
mod_obj = md.to_modem()
```

```python
mod_obj.dataframe.loc[mod_obj.dataframe["t_zx"].values.real == 1.0, "t_zx"] = np.nan
mod_obj.dataframe.loc[mod_obj.dataframe["t_zy"].values.real == 1.0, "t_zy"] = np.nan

mod_obj.dataframe["t_zx_median"] = mod_obj.dataframe.groupby("station")[
    "t_zx"
].transform("median")
mod_obj.dataframe["t_zy_median"] = mod_obj.dataframe.groupby("station")[
    "t_zy"
].transform("median")

mod_obj.dataframe["t_zx_distance"] = abs(
    mod_obj.dataframe["t_zx"] - mod_obj.dataframe["t_zx_median"]
)
mod_obj.dataframe["t_zy_distance"] = abs(
    mod_obj.dataframe["t_zy"] - mod_obj.dataframe["t_zy_median"]
)
mod_obj.dataframe["t_zx_model_error"] = (
    mod_obj.dataframe["t_zx_model_error"]
    + abs(mod_obj.dataframe["t_zx_distance"]) * 0.75
)
mod_obj.dataframe["t_zy_model_error"] = (
    mod_obj.dataframe["t_zy_model_error"]
    + abs(mod_obj.dataframe["t_zy_distance"]) * 0.75
)

mod_obj.dataframe.loc[
    mod_obj.dataframe["t_zx_model_error"] < 0.02, "t_zx_model_error"
] = 0.2
mod_obj.dataframe.loc[
    mod_obj.dataframe["t_zy_model_error"] < 0.02, "t_zy_model_error"
] = 0.2
mod_obj.write_data_file(
    r"c:\Users\jpeacock\OneDrive - DOI\Geothermal\GreatBasin\modem_inv\mina_inv_01\mina_modem_data_z03_t02_tec_05.dat"
)
```
