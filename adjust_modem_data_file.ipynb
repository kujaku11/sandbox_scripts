```python
from mtpy import MT, MTData
from mtpy.modeling.modem import Data
import numpy as np
```

```python
fn = r"c:\Users\jpeacock\OneDrive - DOI\Geothermal\GreatBasin\modem_inv\mina_inv_01\mina_modem_data_z03_t02_tec_07.dat"
```

```python
md = MTData()
md.from_modem(fn)
```

```python
modem_data = Data()
modem_data.dataframe = modem_data.read_data_file(fn)
```

```python
def cap_at_group_mean_real(group_series):
    thresh = 0.4
    out = group_series.copy()
    real = np.real(group_series)
    imag = np.imag(group_series)

    index = np.where(real != np.nan)

    if np.nanmean(real) >= thresh:
        out[index] = real[index] - 0.5 + 1j * imag[index]
    elif np.nanmean(real) <= -1 * thresh:
        out[index] = real[index] + 0.5 + 1j * imag[index]
    return out


modem_data.dataframe["t_zy"] = modem_data.dataframe.groupby("station")[
    "t_zy"
].transform(cap_at_group_mean_real)
```

```python
modem_data.write_data_file(
    file_name=r"c:\Users\jpeacock\OneDrive - DOI\Geothermal\GreatBasin\modem_inv\mina_inv_01\mina_modem_data_z03_t02_tec_07.dat"
)
```
