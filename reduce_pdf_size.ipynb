```python
from pypdf import PdfWriter

new_pdf_fn = pdf_file.parent / f"{pdf_file.stem}_compressed.pdf"
writer = PdfWriter(clone_from=pdf_file)

for page in writer.pages:
    page.compress_content_streams(level=9)
    for img in page.images:
        img.replace(img.image, quality=75)  # re-encode as JPEG at quality 75

writer.compress_identical_objects(remove_duplicates=True, remove_unreferenced=True)

with open(new_pdf_fn, "wb") as f:
    writer.write(f)

print(f"Original File size: {pdf_file.stat().st_size / 1E6} Mb")
print(f"Compressed size:    {new_pdf_fn.stat().st_size / 1E6} Mb")
```

```python
from pathlib import Path
import shutil

fn = Path(r"c:\Users\jpeacock\OneDrive - DOI\ClearLake\jvgr_manuscript\clvf_jgr.tex")

fig_path = fn.parent

with open(fn, "r") as fid:
    lines = fid.readlines()

count = 1
for line in lines:
    if line[0] == "%":
        continue
    if line.lower().find("includegraphics") > 0:
        line_str = line.replace("{", " ").replace("}", " ")
        line_list = line_str.strip().replace(";", "").split()
        # fig_dir_path = os.path.dirname(line_list[-1][1:])
        fig_fn = fig_path.joinpath(line_list[-1])
        numbered_fn = fig_path.joinpath(f"figure_{count:02}.png")
        shutil.copy(fig_fn, numbered_fn)
        count += 1
```
